````python
features = [
    "sell_price",
    "day_of_week",
    "month",
    "year",
    "demand_lag_1",
    "demand_lag_7",
    "rolling_mean_7",
    "rolling_mean_28"
]

ml_data = model_data.dropna(
    subset=features + ["demand"]
)

print(ml_data.shape)
ml_data[features + ["demand"]].head()
````

````python
print("First date:", ml_data["date"].min())
print("Last date:", ml_data["date"].max())
print("Number of rows:", len(ml_data))
````

````python
cutoff_date = ml_data["date"].max() - pd.Timedelta(days=28)

train = ml_data[ml_data["date"] <= cutoff_date]
test = ml_data[ml_data["date"] > cutoff_date]

print("Cutoff:", cutoff_date)
print("Train:", train["date"].min(), "→", train["date"].max())
print("Test:", test["date"].min(), "→", test["date"].max())

print("Train rows:", len(train))
print("Test rows:", len(test))
````

````python
X_train = train[features]
y_train = train["demand"]

X_test = test[features]
y_test = test["demand"]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
````

````python
from sklearn.ensemble import HistGradientBoostingRegressor

model = HistGradientBoostingRegressor(
    max_iter=100,
    learning_rate=0.1,
    max_depth=8,
    random_state=42
)

model.fit(X_train, y_train)

print("Model trained!")
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", round(mae, 3))
print("RMSE:", round(rmse, 3))
print("R²:", round(r2, 3))
````

````python
elasticity_train = model_data[
    model_data["date"] <= cutoff_date
]

candidate_train = (
    elasticity_train.groupby("item_id")
    .agg(
        unique_prices=("sell_price", "nunique"),
        observations=("demand", "size"),
        selling_days=("demand", lambda x: (x > 0).sum())
    )
)

candidate_items = candidate_train[
    (candidate_train["unique_prices"] >= 4) &
    (candidate_train["selling_days"] >= 100)
].index

print("Eligible products:", len(candidate_items))
````

````python
# Select only 25 products once
test_items = list(candidate_items[:25])

elasticity_subset = elasticity_train[
    elasticity_train["item_id"].isin(test_items)
][[
    "item_id",
    "date",
    "demand",
    "sell_price",
    "day_of_week",
    "month",
    "demand_lag_1",
    "demand_lag_7",
    "rolling_mean_7",
    "rolling_mean_28"
]].dropna().copy()

# Create expensive features only once
elasticity_subset["log_price"] = np.log(
    elasticity_subset["sell_price"]
)

elasticity_subset["time_idx"] = (
    elasticity_subset["date"] -
    elasticity_subset["date"].min()
).dt.days

print("Rows to process:", len(elasticity_subset))
print("Products:", elasticity_subset["item_id"].nunique())
````

````python
elasticity_results = []

for i, (item_id, df) in enumerate(
    elasticity_subset.groupby("item_id"), 1
):

    if df["sell_price"].nunique() < 4:
        continue

    temp = pd.get_dummies(
        df,
        columns=["day_of_week", "month"],
        drop_first=True,
        dtype=float
    )

    X = temp.drop(columns=[
        "item_id",
        "date",
        "demand",
        "sell_price"
    ])

    y = temp["demand"]

    model_e = PoissonRegressor(
        alpha=0.001,
        max_iter=300
    )

    try:
        model_e.fit(X, y)

        elasticity = model_e.coef_[
            X.columns.get_loc("log_price")
        ]

        elasticity_results.append({
            "item_id": item_id,
            "elasticity": elasticity,
            "avg_price": df["sell_price"].mean(),
            "avg_demand": y.mean(),
            "observations": len(df),
            "unique_prices": df["sell_price"].nunique()
        })

        print(f"{i}/25 done — {item_id}")

    except Exception as e:
        print(f"{i}/25 failed — {item_id}: {e}")
````

````python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import PoissonRegressor
import numpy as np
import pandas as pd

elasticity_results = []

scale_cols = [
    "demand_lag_1",
    "demand_lag_7",
    "rolling_mean_7",
    "rolling_mean_28",
    "time_idx"
]

for i, (item_id, df) in enumerate(
    elasticity_subset.groupby("item_id"), 1
):

    if df["sell_price"].nunique() < 4:
        continue

    temp = pd.get_dummies(
        df,
        columns=["day_of_week", "month"],
        drop_first=True,
        dtype=float
    )

    # Remove invalid values
    temp = temp.replace([np.inf, -np.inf], np.nan).dropna()

    # Scale controls, NOT log_price
    scaler = StandardScaler()
    temp[scale_cols] = scaler.fit_transform(temp[scale_cols])

    X = temp.drop(columns=[
        "item_id",
        "date",
        "demand",
        "sell_price"
    ])

    y = temp["demand"]

    model_e = PoissonRegressor(
        alpha=0.01,
        max_iter=1000,
        tol=1e-6
    )

    try:
        model_e.fit(X, y)

        elasticity = model_e.coef_[
            X.columns.get_loc("log_price")
        ]

        elasticity_results.append({
            "item_id": item_id,
            "elasticity": elasticity,
            "avg_price": df["sell_price"].mean(),
            "avg_demand": df["demand"].mean(),
            "observations": len(df),
            "unique_prices": df["sell_price"].nunique(),
            "iterations": model_e.n_iter_
        })

        print(
            f"{i}/25 done — {item_id} "
            f"| elasticity={elasticity:.3f} "
            f"| iterations={model_e.n_iter_}"
        )

    except Exception as e:
        print(f"{i}/25 FAILED — {item_id}: {e}")
````

````python
elasticity_table = pd.DataFrame(elasticity_results)

def classify_elasticity(e):
    if e <= -1:
        return "Elastic"
    elif e < 0:
        return "Inelastic"
    else:
        return "Investigate"

elasticity_table["classification"] = (
    elasticity_table["elasticity"]
    .apply(classify_elasticity)
)

elasticity_table.sort_values("elasticity")
````

````python
all_items = list(candidate_items)

batch_size = 100
all_elasticity_results = []

print("Products to process:", len(all_items))
````

````python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import PoissonRegressor

for start in range(0, len(all_items), batch_size):

    batch_items = all_items[start:start + batch_size]

    batch = elasticity_train[
        elasticity_train["item_id"].isin(batch_items)
    ][[
        "item_id",
        "date",
        "demand",
        "sell_price",
        "day_of_week",
        "month",
        "demand_lag_1",
        "demand_lag_7",
        "rolling_mean_7",
        "rolling_mean_28"
    ]].dropna().copy()

    batch["log_price"] = np.log(batch["sell_price"])

    batch["time_idx"] = (
        batch["date"] - batch["date"].min()
    ).dt.days

    for item_id, df in batch.groupby("item_id"):

        temp = pd.get_dummies(
            df,
            columns=["day_of_week", "month"],
            drop_first=True,
            dtype=float
        )

        temp = temp.replace(
            [np.inf, -np.inf],
            np.nan
        ).dropna()

        scaler = StandardScaler()

        temp[scale_cols] = scaler.fit_transform(
            temp[scale_cols]
        )

        X = temp.drop(columns=[
            "item_id",
            "date",
            "demand",
            "sell_price"
        ])

        y = temp["demand"]

        model_e = PoissonRegressor(
            alpha=0.01,
            max_iter=1000,
            tol=1e-6
        )

        try:
            model_e.fit(X, y)

            e = model_e.coef_[
                X.columns.get_loc("log_price")
            ]

            all_elasticity_results.append({
                "item_id": item_id,
                "elasticity": e,
                "avg_price": df["sell_price"].mean(),
                "avg_demand": df["demand"].mean(),
                "observations": len(df),
                "unique_prices": df["sell_price"].nunique(),
                "iterations": model_e.n_iter_
            })

        except Exception as error:
            print("FAILED:", item_id, error)

    # save progress after each batch
    pd.DataFrame(
        all_elasticity_results
    ).to_csv(
        "data/elasticity_results_checkpoint.csv",
        index=False
    )

    print(
        f"Completed {min(start + batch_size, len(all_items))}"
        f"/{len(all_items)}"
    )
````

````python

def recommend_price(item_id, min_price_days=14):

    # 1. Get product elasticity
    row = master_elasticity[
        master_elasticity["item_id"] == item_id
    ]

    if row.empty:
        return {
            "item_id": item_id,
            "status": "NO_ELASTICITY_MODEL"
        }

    elasticity = row["elasticity"].iloc[0]
    classification = row["classification"].iloc[0]

    # Governance rule
    if classification == "Investigate":
        return {
            "item_id": item_id,
            "status": "MANUAL_REVIEW",
            "elasticity": elasticity,
            "classification": classification
        }

    # 2. Get latest available test observation
    product_test = test[
        test["item_id"] == item_id
    ].sort_values("date")

    if product_test.empty:
        return {
            "item_id": item_id,
            "status": "NO_TEST_OBSERVATION"
        }

    latest = product_test.iloc[[-1]]

    base_price = latest["sell_price"].iloc[0]

    # 3. Baseline demand forecast
    base_forecast = max(
        model.predict(latest[features])[0],
        0
    )

    # 4. Only use prices with enough historical support
    price_support = (
        elasticity_train[
            elasticity_train["item_id"] == item_id
        ]
        .groupby("sell_price")
        .size()
    )

    safe_prices = price_support[
        price_support >= min_price_days
    ].index.to_numpy()

    if len(safe_prices) < 2:
        return {
            "item_id": item_id,
            "status": "INSUFFICIENT_PRICE_HISTORY",
            "safe_prices": len(safe_prices)
        }

    # 5. Simulate prices
    scenarios = []

    for new_price in safe_prices:

        adjusted_demand = base_forecast * (
            new_price / base_price
        ) ** elasticity

        adjusted_demand = max(adjusted_demand, 0)

        expected_revenue = (
            new_price * adjusted_demand
        )

        scenarios.append({
            "price": new_price,
            "forecast_demand": adjusted_demand,
            "expected_revenue": expected_revenue
        })

    scenarios = pd.DataFrame(scenarios)

    best = scenarios.loc[
        scenarios["expected_revenue"].idxmax()
    ]

    current_revenue = (
        base_price * base_forecast
    )

    uplift_pct = (
        (best["expected_revenue"] - current_revenue)
        / current_revenue * 100
        if current_revenue > 0
        else 0
    )

    return {
        "item_id": item_id,
        "status": "AUTO_RECOMMEND",
        "classification": classification,
        "elasticity": round(elasticity, 3),
        "current_price": round(base_price, 3),
        "recommended_price": round(best["price"], 3),
        "forecast_current_demand": round(base_forecast, 3),
        "forecast_recommended_demand": round(
            best["forecast_demand"], 3
        ),
        "current_expected_revenue": round(
            current_revenue, 3
        ),
        "recommended_expected_revenue": round(
            best["expected_revenue"], 3
        ),
        "expected_revenue_uplift_pct": round(
            uplift_pct, 2
        )
    }
````

````python
from pathlib import Path

path = Path("src/pricing_engine.py")

text = path.read_text(encoding="utf-8")

v2_code = r'''

# ============================================================
# BACKEND V2 — AUTOMATIC PRICING RECOMMENDATION
# ============================================================

FEATURE_STORE_PATH = (
    BASE_DIR
    / "data"
    / "latest_product_features.csv"
)

latest_product_features = pd.read_csv(
    FEATURE_STORE_PATH
)


def get_latest_product_features(item_id):
    """
    Retrieve the most recent ML feature snapshot
    available for a product.
    """

    row = latest_product_features[
        latest_product_features["item_id"] == item_id
    ]

    if row.empty:
        return None

    row = row.iloc[0]

    return {
        "item_id": item_id,
        "date": str(row["date"]),
        "sell_price": float(row["sell_price"]),
        "day_of_week": float(row["day_of_week"]),
        "month": float(row["month"]),
        "year": float(row["year"]),
        "demand_lag_1": float(row["demand_lag_1"]),
        "demand_lag_7": float(row["demand_lag_7"]),
        "rolling_mean_7": float(row["rolling_mean_7"]),
        "rolling_mean_28": float(row["rolling_mean_28"])
    }


def generate_candidate_prices(
    base_price,
    max_change_pct=0.10
):
    """
    Generate controlled candidate prices around
    the current price.

    Default range:
    -10%, -5%, current, +5%, +10%
    """

    multipliers = [
        1 - max_change_pct,
        1 - max_change_pct / 2,
        1.0,
        1 + max_change_pct / 2,
        1 + max_change_pct
    ]

    prices = [
        round(
            max(base_price * m, 0.01),
            2
        )
        for m in multipliers
    ]

    return sorted(set(prices))


def validate_price_guardrails(
    base_price,
    candidate_prices,
    max_change_pct=0.10
):
    """
    Prevent prices outside the allowed
    percentage change around the current price.
    """

    minimum_price = (
        base_price
        * (1 - max_change_pct)
    )

    maximum_price = (
        base_price
        * (1 + max_change_pct)
    )

    approved = []

    for price in candidate_prices:

        if (
            price >= minimum_price - 0.001
            and
            price <= maximum_price + 0.001
            and
            price > 0
        ):
            approved.append(
                round(float(price), 2)
            )

    return sorted(set(approved))


def recommend_price_auto(
    item_id,
    max_change_pct=0.10
):
    """
    Fully automatic pricing workflow.

    Product ID
        ↓
    Feature store
        ↓
    Demand forecast
        ↓
    Elasticity lookup
        ↓
    Governance
        ↓
    Candidate price generation
        ↓
    Guardrails
        ↓
    Revenue optimization
    """

    # ------------------------------------------
    # 1. Product features
    # ------------------------------------------

    snapshot = get_latest_product_features(
        item_id
    )

    if snapshot is None:

        return {
            "status": "NOT_FOUND",
            "item_id": item_id,
            "reason": (
                "Latest product features "
                "not available."
            )
        }


    # ------------------------------------------
    # 2. Elasticity
    # ------------------------------------------

    elasticity_info = get_elasticity(
        item_id
    )

    if elasticity_info is None:

        return {
            "status": "NOT_FOUND",
            "item_id": item_id,
            "reason": (
                "Product elasticity "
                "not available."
            )
        }


    # ------------------------------------------
    # 3. Automatic ML demand forecast
    # ------------------------------------------

    feature_values = {
        feature:
        snapshot[feature]
        for feature in features
    }

    predicted_demand = forecast_demand(
        feature_values
    )

    current_price = float(
        snapshot["sell_price"]
    )


    # ------------------------------------------
    # 4. Human-in-the-loop governance
    # ------------------------------------------

    if (
        elasticity_info[
            "classification"
        ]
        ==
        "Investigate"
    ):

        return {
            "status": "MANUAL_REVIEW",
            "item_id": item_id,
            "current_price": current_price,
            "forecast_demand": predicted_demand,
            "elasticity": elasticity_info[
                "elasticity"
            ],
            "classification": elasticity_info[
                "classification"
            ],
            "reason": (
                "Elasticity estimate requires "
                "human validation before repricing."
            )
        }


    # ------------------------------------------
    # 5. Automatic candidate prices
    # ------------------------------------------

    candidate_prices = (
        generate_candidate_prices(
            current_price,
            max_change_pct
        )
    )


    # ------------------------------------------
    # 6. Business guardrails
    # ------------------------------------------

    approved_prices = (
        validate_price_guardrails(
            current_price,
            candidate_prices,
            max_change_pct
        )
    )


    if not approved_prices:

        return {
            "status": "MANUAL_REVIEW",
            "item_id": item_id,
            "reason": (
                "No candidate price passed "
                "business guardrails."
            )
        }


    # ------------------------------------------
    # 7. Revenue optimization
    # ------------------------------------------

    optimization = optimize_price(
        base_price=current_price,
        base_demand=predicted_demand,
        elasticity=elasticity_info[
            "elasticity"
        ],
        candidate_prices=approved_prices
    )


    current_revenue = (
        current_price
        *
        predicted_demand
    )

    recommended_revenue = (
        optimization[
            "expected_revenue"
        ]
    )

    revenue_uplift_pct = (
        (
            recommended_revenue
            -
            current_revenue
        )
        /
        current_revenue
        *
        100
        if current_revenue > 0
        else 0.0
    )


    return {
        "status": "AUTO_RECOMMEND",
        "item_id": item_id,

        "snapshot_date":
        snapshot["date"],

        "current_price":
        current_price,

        "forecast_demand":
        predicted_demand,

        "elasticity":
        elasticity_info[
            "elasticity"
        ],

        "classification":
        elasticity_info[
            "classification"
        ],

        "candidate_prices":
        approved_prices,

        "recommended_price":
        optimization[
            "recommended_price"
        ],

        "forecast_demand_at_recommended_price":
        optimization[
            "forecast_demand"
        ],

        "current_expected_revenue":
        current_revenue,

        "recommended_expected_revenue":
        recommended_revenue,

        "expected_revenue_uplift_pct":
        revenue_uplift_pct,

        "guardrail_max_change_pct":
        max_change_pct * 100,

        "scenarios":
        optimization[
            "scenarios"
        ]
    }
'''

if "def recommend_price_auto(" not in text:
    text += v2_code

    path.write_text(
        text,
        encoding="utf-8"
    )

    print("Backend V2 added successfully!")

else:
    print("Backend V2 already exists — nothing changed.")
````

````python
from pathlib import Path

path = Path("src/pricing_engine.py")
text = path.read_text(encoding="utf-8")

profit_code = r'''

# ============================================================
# BACKEND V3 — PROFIT OPTIMIZATION
# ============================================================

def optimize_profit(
    base_price,
    base_demand,
    elasticity,
    candidate_prices,
    unit_cost
):
    """
    Compare candidate prices using both expected
    revenue and expected gross profit.

    Profit = (Price - Unit Cost) * Forecast Demand
    """

    if unit_cost < 0:
        raise ValueError(
            "unit_cost cannot be negative."
        )

    scenarios = []

    for price in candidate_prices:

        demand = (
            base_demand
            *
            (price / base_price)
            ** elasticity
        )

        demand = max(
            float(demand),
            0.0
        )

        revenue = float(
            price * demand
        )

        profit = float(
            (price - unit_cost)
            * demand
        )

        margin_pct = (
            (
                price - unit_cost
            )
            /
            price
            *
            100
            if price > 0
            else 0.0
        )

        scenarios.append({
            "price": float(price),
            "forecast_demand": demand,
            "expected_revenue": revenue,
            "expected_profit": profit,
            "gross_margin_pct": margin_pct
        })


    revenue_best = max(
        scenarios,
        key=lambda x:
        x["expected_revenue"]
    )

    profit_best = max(
        scenarios,
        key=lambda x:
        x["expected_profit"]
    )


    current_profit = (
        (base_price - unit_cost)
        *
        base_demand
    )

    profit_uplift_pct = (
        (
            profit_best["expected_profit"]
            -
            current_profit
        )
        /
        abs(current_profit)
        *
        100
        if current_profit != 0
        else 0.0
    )


    return {
        "unit_cost": float(unit_cost),

        "revenue_maximizing_price":
        revenue_best["price"],

        "profit_maximizing_price":
        profit_best["price"],

        "expected_revenue_at_revenue_optimum":
        revenue_best["expected_revenue"],

        "expected_profit_at_profit_optimum":
        profit_best["expected_profit"],

        "forecast_demand_at_profit_optimum":
        profit_best["forecast_demand"],

        "current_expected_profit":
        float(current_profit),

        "expected_profit_uplift_pct":
        float(profit_uplift_pct),

        "scenarios":
        scenarios
    }
'''

if "def optimize_profit(" not in text:

    text += profit_code

    path.write_text(
        text,
        encoding="utf-8"
    )

    print("Profit optimization engine added successfully!")

else:

    print("Profit optimization already exists.")
````

````python
from pathlib import Path

path = Path("src/pricing_engine.py")
text = path.read_text(encoding="utf-8")

explanation_code = r'''

# ============================================================
# BACKEND V4 — EXECUTIVE DECISION EXPLANATION
# ============================================================

def build_pricing_explanation(result):
    """
    Convert pricing-engine outputs into a concise
    executive decision explanation and governance checks.
    """

    if result.get("status") != "AUTO_RECOMMEND":

        return {
            "decision_summary": (
                result.get(
                    "reason",
                    "Recommendation requires human review."
                )
            ),
            "governance_checks": {
                "automatic_recommendation": False
            }
        }


    current_price = float(
        result["current_price"]
    )

    recommended_price = float(
        result["profit_maximizing_price"]
    )

    elasticity = float(
        result["elasticity"]
    )

    classification = result[
        "classification"
    ]

    profit_uplift = float(
        result["expected_profit_uplift_pct"]
    )

    max_change = float(
        result["guardrail_max_change_pct"]
    )

    unit_cost = float(
        result["unit_cost"]
    )


    # ------------------------------------------
    # Price direction
    # ------------------------------------------

    if recommended_price > current_price:

        direction_text = (
            f"Increase the price from "
            f"{current_price:.2f} to "
            f"{recommended_price:.2f}."
        )

    elif recommended_price < current_price:

        direction_text = (
            f"Decrease the price from "
            f"{current_price:.2f} to "
            f"{recommended_price:.2f}."
        )

    else:

        direction_text = (
            f"Maintain the current price at "
            f"{current_price:.2f}."
        )


    # ------------------------------------------
    # Elasticity interpretation
    # ------------------------------------------

    if classification == "Inelastic":

        elasticity_text = (
            "The SKU is relatively price-inelastic, "
            "so forecast demand is expected to remain "
            "comparatively stable under moderate price changes."
        )

    elif classification == "Elastic":

        elasticity_text = (
            "The SKU is price-sensitive, so demand may "
            "respond materially to price changes."
        )

    else:

        elasticity_text = (
            "The estimated price-response relationship "
            "requires additional validation."
        )


    # ------------------------------------------
    # Profit interpretation
    # ------------------------------------------

    if profit_uplift > 0:

        profit_text = (
            f"The selected price is expected to improve "
            f"gross profit by approximately "
            f"{profit_uplift:.1f}%."
        )

    elif profit_uplift < 0:

        profit_text = (
            f"The scenario implies an expected gross-profit "
            f"change of {profit_uplift:.1f}%."
        )

    else:

        profit_text = (
            "The selected scenario does not materially "
            "change expected gross profit."
        )


    guardrail_text = (
        f"The recommendation remains inside the configured "
        f"±{max_change:.0f}% pricing guardrail."
    )


    decision_summary = " ".join(
        [
            direction_text,
            elasticity_text,
            profit_text,
            guardrail_text
        ]
    )


    # ------------------------------------------
    # Governance checks
    # ------------------------------------------

    price_change_pct = (
        (
            recommended_price
            -
            current_price
        )
        /
        current_price
        *
        100
        if current_price > 0
        else 0
    )


    governance_checks = {
        "elasticity_accepted":
        classification in [
            "Elastic",
            "Inelastic"
        ],

        "automatic_recommendation":
        result["status"]
        ==
        "AUTO_RECOMMEND",

        "price_within_guardrails":
        abs(price_change_pct)
        <=
        max_change + 0.01,

        "price_above_unit_cost":
        recommended_price
        >
        unit_cost,

        "positive_expected_profit":
        result[
            "expected_profit_at_profit_optimum"
        ]
        >
        0
    }


    return {
        "decision_summary":
        decision_summary,

        "governance_checks":
        governance_checks
    }
'''

if "def build_pricing_explanation(" not in text:

    text += explanation_code

    path.write_text(
        text,
        encoding="utf-8"
    )

    print(
        "Executive Decision Explanation engine added successfully!"
    )

else:

    print(
        "Decision Explanation engine already exists."
    )
````

````python
from pathlib import Path
import shutil

desktop = Path.home() / "Desktop"

# Correct project destinations
screenshots_target = Path("assets/screenshots")
architecture_target = Path("assets/architecture")

screenshots_target.mkdir(parents=True, exist_ok=True)
architecture_target.mkdir(parents=True, exist_ok=True)

# Your accidentally-created Desktop folders
screenshots_source = desktop / "CUsersiliassoAI_Demand_Pricing_Engineassetsscreenshots"
architecture_source = desktop / "system_architecture.png"

# Copy the four screenshots
expected = [
    "01_executive_overview.png",
    "02_pricing_decision.png",
    "03_portfolio_radar.png",
    "04_model_intelligence.png",
]

for filename in expected:
    src = screenshots_source / filename
    dst = screenshots_target / filename

    if src.exists():
        shutil.copy2(src, dst)
        print("✅ Copied:", filename)
    else:
        print("❌ Missing:", src)

# Copy the architecture image
architecture_files = [
    p for p in architecture_source.iterdir()
    if p.is_file() and p.suffix.lower() == ".png"
]

if architecture_files:
    shutil.copy2(
        architecture_files[0],
        architecture_target / "system_architecture.png"
    )
    print("✅ Copied: system_architecture.png")
else:
    print("❌ Architecture image not found")

print("\nDONE")
````
